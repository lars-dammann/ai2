# Preprocessing of data
This file details the preprocessing of the data before the training of the U-Net

## Image confusions
In the original data the naming of multiple folderes was confused with regard to the "before" and "after" images. This is easily visible from visual inspection of the before and after images.
Specifically, the correct folder correspondence is:

before/2-5-10 <-> after/2-5-11
before/2-5-11 <-> after/2-5-10
before/3-8-3 <-> after/3-8-6
before/3-8-6 <-> after/3-8-3

Hence, one of the folder has to be corrected. In the data I chose to rename the after folders, so they align with the before folders. For the training it is not relevant, however, to connect them to the final corrosion inhibitor in the provided list, the real correspondence between the images and inhibitor still needs to be clarified.

Additionllay, I noticed that two images were identicial, i.e.

before/4-6-11/3.png = before/4-6-10/3.png


From visual inspection it became clear that after/4-6-10/3.png is missing its correct "before" image. Consequently, the images before/4-6-10/3.png and after/4-6-10/3.png were removed from the dataset.


## Renaming of images
To facilitate the work with the data, it was renamed and moved to a single folder. The files received a unqiue ident by concatenating the subfolder names, e.g. before/profilometer/x-x-x/y.png became before/profilometer/x-x-x-y.png
This was archived with the bashscripts/reorder-images.sh script

## Image correspondence before after and profilometer image to height image
To predict the corrosion with pixel precision it is important to ensure a precise overlap of the images. However, multiple potential problems have been discovered with regard to the alignment of the images.
1. There are before images that do not possess after images and vice versa. These files are listed in notebooks/check-data-quality.ipynb
2. For some samples the images and height measurements do not have the same image shape. Consequently, it is difficult to precisely overlap these two informations. In general it was found that the height profil of the samples does often exhibit image borders filled with NaN values. The borders could lead to the image and height data being shifted. While effort was put into precisely overlapping the image and height profiles it can not be guaranteed, that they are pixel precise. For future measurements care should be taken that these two data sources are aligned as well as possible.
3. Many of the before and after images did not exhibit the same image shapes. In general it was found, that the images did not align but the images were shifted against each other. Consequently, the images had to to be overlapped by template matching in notebooks/match-before-after.ipynb. However, during template matching it became clear that many of the images where slightly warped. As a result, while the template matching in general found a good overlap, at some positions, it was not possible to archive a pixel precise overlap over the whole sample. For future measurements, this process should be improved. In particular, marker should be added to the samples, such that the samples can be easily overlapped by template matching. The best would be a visible limitation of the circular sample region, as for low corrosion events, the sample can not distingishued from the background. Additionally, care should be taken as to not change the distance and angle of the camera to facilitate the overlapping of the before and after images
